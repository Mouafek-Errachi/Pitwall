### Data reation and initial inspection

```python
import pandas as pd

data = {
    "Driver": ["VER", "VER", "VER", "HAM", "HAM", "HAM", "LEC", "LEC"],
    "LapNumber": [1, 2, 3, 1, 2, 3, 1, 2],
    "LapTime": [90.1, 90.4, 91.2, 90.8, 89.9, 90.3, 91.5, 90.6],
    "Compound": ["SOFT","SOFT","SOFT","MEDIUM","MEDIUM","MEDIUM","HARD","HARD"],
    "PitStop": [False, False, True, False, False, False, False, True]
}

df = pd.DataFrame(data)
print(df.shape)
print(df.dtypes)
print(df.head())
print(df.describe())
print(df.isnull().sum())
print(df.nunique())
```

### Filter pit stops and best lap per driver

```python
df1= df[df["PitStop"]==False]
df2 = df1.groupby("Driver")["LapTime"].min()
df2 = df2.sort_values( ascending=True)
print(df1)
print(df2)
```

### LapTime filtering and iloc, and loc difference

```python
df_filtered = df1[df1["LapTime"] < 91]
print(df_filtered)
print()
print(df_filtered.iloc[1])
print()
print(df_filtered.loc[1])
df_reset = df_filtered.set_index(pd.Index([1,5,3,4,0]))
print()
print(df_reset.iloc[1])
print()
print(df_reset.loc[1])
```

### Dirty data conversion & coercion

```python
dirty_data = {
    "Driver": ["VER", "HAM", "LEC", "VER"],
    "Speed": [287.3, "n/a", 301.1, 295.7]
}

df_dirty = pd.DataFrame(dirty_data)
print(df_dirty)
df_dirty["Speed"] = pd.to_numeric(df_dirty["Speed"], errors="coerce")
print(df_dirty)
assert all(t is pd.NA for t in df_dirty["Speed"]), "There are still NaN values in the Speed column"
```

### Format lap times and add formatted column

```python
def format_laptime(seconds):
    minutes = int(seconds // 60)
    secs = seconds % 60
    return f"{minutes}:{secs:06.3f}"

df["LapTimeFormatted"] = df["LapTime"].apply(format_laptime)
# Only use .apply() if the number of rows is small. For large datasets, use vectorized operations instead.
print(df)
```
