### Data reation and initial inspection

```python
import pandas as pd

data = {
    "Driver": ["VER", "VER", "VER", "HAM", "HAM", "HAM", "LEC", "LEC"],
    "LapNumber": [1, 2, 3, 1, 2, 3, 1, 2],
    "LapTime": [90.1, 90.4, 91.2, 90.8, 89.9, 90.3, 91.5, 90.6],
    "Compound": ["SOFT","SOFT","SOFT","MEDIUM","MEDIUM","MEDIUM","HARD","HARD"],
    "PitStop": [False, False, True, False, False, False, False, True]
}

df = pd.DataFrame(data)
print(df.shape)
print(df.dtypes)
print(df.head())
print(df.describe())
print(df.isnull().sum())
print(df["Compound"].unique())
print(df["Driver"].unique())
```

### Filter pit stops and best lap per driver

```python
df1= df[df["PitStop"]==False]
df2 = df1.groupby("Driver")["LapTime"].min()
df2 = df2.sort_values( ascending=True)
print(df1)
print(df2)
```

### LapTime filtering and iloc, and loc difference

```python
df_filtered = df1[df1["LapTime"] < 91]
print(df_filtered)
print()
print(df_filtered.iloc[3])
print()
print(df_filtered.loc[3])
print()
df_reset = df_filtered.reset_index(drop=True)
print(df_reset)
print()
print(df_reset.iloc[3])
print()
print(df_reset.loc[3])
```

The functions iloc and loc give two different results:
- iloc gives the results at integer location.
- loc gives the results according to the labeling index.

After the reset, they became identical at the same index.

### Dirty data conversion & coercion

```python
dirty_data = {
    "Driver": ["VER", "HAM", "LEC", "HAM"],
    "Speed": [247.3, "n/a", 251.1, 245.7]
}

df_dirty = pd.DataFrame(dirty_data)
print(df_dirty)
df_dirty["Speed"] = pd.to_numeric(df_dirty["Speed"], errors="coerce")
print(df_dirty)
assert df_dirty["Speed"].isna().sum() == 1, "Expected exactly 1 NaN from coercion"
assert df_dirty["Speed"].dropna().max() < 260, f"Impossible speed: {df_dirty['Speed'].max()} km/h"
assert df_dirty["Speed"].dropna().min() > 0, f"Impossible speed: {df_dirty['Speed'].min()} km/h, must be positive"
```

### Format lap times and add formatted column

```python
def format_laptime(seconds):
    minutes = int(seconds // 60)
    secs = seconds % 60
    return f"{minutes}:{secs:06.3f}"

df["LapTimeFormatted"] = df["LapTime"].apply(format_laptime)
# Only use .apply() if the number of rows is small. For large datasets, use vectorized operations instead.
print(df)
```
